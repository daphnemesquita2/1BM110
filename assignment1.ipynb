```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv('energy_generation_train.csv')

# Ensure all columns have the correct data type
df['DateTime'] = pd.to_datetime(df['DateTime'])
numeric_cols = df.columns.drop('DateTime')
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

cols_to_drop = df.columns[df.fillna(0).eq(0).all()].tolist()

# Drop the columns
df = df.drop(columns=cols_to_drop)
df.head()
```

```python
# Basic statistics
stats = df.describe()

print(f"Dropped columns: {cols_to_drop}")
print("\nRemaining columns:", df.columns.tolist())
stats
```
