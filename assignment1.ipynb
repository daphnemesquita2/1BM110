```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv('energy_generation_train.csv')

# Ensure all columns have the correct data type
df['DateTime'] = pd.to_datetime(df['DateTime'])
numeric_cols = df.columns.drop('DateTime')
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# Identify columns where all values are either NaN OR Zero
# We fill NaNs with 0 temporarily; if the whole column is then 0, we drop it.
cols_to_drop = df.columns[df.fillna(0).eq(0).all()].tolist()

# Drop the columns
df = df.drop(columns=cols_to_drop)

# Basic statistics
stats = df.describe()

print(f"Dropped columns: {cols_to_drop}")
print("\nRemaining columns:", df.columns.tolist())
stats
```

```python
# 4. Describe basic statistics of the numerical variables
# This includes count, mean, std, min, 25%, 50%, 75%, and max
basic_stats = df.describe()

# Output results for inspection
print(f"Columns removed: {cols_to_drop}")
print("\nBasic Statistics:")
print(basic_stats)
```
